```python
import os

train_dir = '/content/blood_cells_data/dataset2-master/dataset2-master/images/TRAIN'
test_dir = '/content/blood_cells_data/dataset2-master/dataset2-master/images/TEST'

print("TRAIN counts:")
for cls in os.listdir(train_dir):
    count = len(os.listdir(os.path.join(train_dir, cls)))
    print(f"  {cls}: {count} images")

print("\nTEST counts:")
for cls in os.listdir(test_dir):
    count = len(os.listdir(os.path.join(test_dir, cls)))
    print(f"  {cls}: {count} images")
```

```python
import torch
import torch.nn as nn
from torchvision import datasets, transforms, models
from torch.utils.data import DataLoader

# Image transforms: resize, normalize (standard ImageNet stats since we're using a pretrained model)
train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

test_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

train_dataset = datasets.ImageFolder(train_dir, transform=train_transform)
test_dataset = datasets.ImageFolder(test_dir, transform=test_transform)

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True, num_workers=2)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False, num_workers=2)

print("Classes:", train_dataset.classes)
print("Train samples:", len(train_dataset))
print("Test samples:", len(test_dataset))
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

model = models.resnet18(weights='IMAGENET1K_V1')

# Replace the final layer to output 4 classes instead of 1000
num_features = model.fc.in_features
model.fc = nn.Linear(num_features, 4)

model = model.to(device)

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

print("Model ready!")
```

```python
import time

num_epochs = 5

for epoch in range(num_epochs):
    start = time.time()
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()
        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

    epoch_loss = running_loss / len(train_loader)
    epoch_acc = 100 * correct / total
    elapsed = time.time() - start
    print(f"Epoch {epoch+1}/{num_epochs} — Loss: {epoch_loss:.4f} — Train Accuracy: {epoch_acc:.2f}% — Time: {elapsed:.1f}s")

print("Training complete!")
```

```python
model.eval()
correct = 0
total = 0

with torch.no_grad():
    for images, labels in test_loader:
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

test_accuracy = 100 * correct / total
print(f"Test Accuracy: {test_accuracy:.2f}%")
```

```python
from sklearn.metrics import classification_report, confusion_matrix
import numpy as np

model.eval()
all_preds = []
all_labels = []

with torch.no_grad():
    for images, labels in test_loader:
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        _, predicted = torch.max(outputs, 1)
        all_preds.extend(predicted.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())

print(classification_report(all_labels, all_preds, target_names=train_dataset.classes))
print("\nConfusion Matrix:")
print(confusion_matrix(all_labels, all_preds))
```

```python
# Rebuild the model with a frozen backbone + dropout, and add a validation split

from torch.utils.data import random_split

# Split train into train/val (90/10) so we can monitor real generalization during training
val_size = int(0.1 * len(train_dataset))
train_size = len(train_dataset) - val_size
train_subset, val_subset = random_split(train_dataset, [train_size, val_size])

train_loader = DataLoader(train_subset, batch_size=32, shuffle=True, num_workers=2)
val_loader = DataLoader(val_subset, batch_size=32, shuffle=False, num_workers=2)

# Rebuild model: freeze all layers except the last block and the new classifier head
model = models.resnet18(weights='IMAGENET1K_V1')

for param in model.parameters():
    param.requires_grad = False

for param in model.layer4.parameters():  # unfreeze only the last block
    param.requires_grad = True

num_features = model.fc.in_features
model.fc = nn.Sequential(
    nn.Dropout(0.4),
    nn.Linear(num_features, 4)
)

model = model.to(device)

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(filter(lambda p: p.requires_grad, model.parameters()), lr=0.0005)

print("Rebuilt model with frozen backbone + dropout. Train:", train_size, "Val:", val_size)
```

```python
num_epochs = 8
best_val_acc = 0

for epoch in range(num_epochs):
    # Training
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()
        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

    train_loss = running_loss / len(train_loader)
    train_acc = 100 * correct / total

    # Validation
    model.eval()
    val_correct = 0
    val_total = 0
    with torch.no_grad():
        for images, labels in val_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            _, predicted = torch.max(outputs, 1)
            val_total += labels.size(0)
            val_correct += (predicted == labels).sum().item()

    val_acc = 100 * val_correct / val_total

    print(f"Epoch {epoch+1}/{num_epochs} — Train Loss: {train_loss:.4f} — Train Acc: {train_acc:.2f}% — Val Acc: {val_acc:.2f}%")

    if val_acc > best_val_acc:
        best_val_acc = val_acc
        torch.save(model.state_dict(), 'best_model.pth')
        print(f"  -> New best model saved (Val Acc: {val_acc:.2f}%)")

print(f"\nBest validation accuracy: {best_val_acc:.2f}%")
```

```python
import torch.nn.functional as F

# Define several augmented views for test-time augmentation
tta_transforms = [
    transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ]),
    transforms.Compose([
        transforms.Resize((256, 256)),
        transforms.CenterCrop(224),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ]),
    transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.RandomHorizontalFlip(p=1.0),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ]),
]

model.load_state_dict(torch.load('best_model_v3.pth'))
model.eval()

test_dataset_paths = datasets.ImageFolder(test_dir)  # just to get file paths + labels
all_preds = []
all_labels = []

with torch.no_grad():
    for path, label in test_dataset_paths.samples:
        img = Image.open(path).convert('RGB')
        probs_sum = torch.zeros(1, 4).to(device)

        for t in tta_transforms:
            img_t = t(img).unsqueeze(0).to(device)
            outputs = model(img_t)
            probs_sum += F.softmax(outputs, dim=1)

        avg_probs = probs_sum / len(tta_transforms)
        pred = torch.argmax(avg_probs, dim=1).item()

        all_preds.append(pred)
        all_labels.append(label)

test_accuracy = 100 * sum(p == l for p, l in zip(all_preds, all_labels)) / len(all_labels)
print(f"TTA Test Accuracy: {test_accuracy:.2f}%")
print("\n", classification_report(all_labels, all_preds, target_names=train_dataset.classes))
```

```python
def remove_black_border(img, threshold=10):
    """Crop out black border regions from an image."""
    gray = np.array(img.convert('L'))
    mask = gray > threshold  # True where pixel is NOT black

    rows = np.any(mask, axis=1)
    cols = np.any(mask, axis=0)

    if not rows.any() or not cols.any():
        return img  # fully black, edge case — return as-is

    rmin, rmax = np.where(rows)[0][[0, -1]]
    cmin, cmax = np.where(cols)[0][[0, -1]]

    return img.crop((cmin, rmin, cmax + 1, rmax + 1))

# Test it on a few sample images to confirm it works well
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
sample_classes = ['EOSINOPHIL', 'LYMPHOCYTE', 'MONOCYTE', 'NEUTROPHIL']

for i, cls in enumerate(sample_classes):
    cls_dir = os.path.join(train_dir, cls)
    f = os.listdir(cls_dir)[0]
    img = Image.open(os.path.join(cls_dir, f)).convert('RGB')
    cropped = remove_black_border(img)

    axes[0, i].imshow(img)
    axes[0, i].set_title(f"{cls} — Original")
    axes[0, i].axis('off')

    axes[1, i].imshow(cropped)
    axes[1, i].set_title(f"{cls} — Border Removed")
    axes[1, i].axis('off')

plt.tight_layout()
plt.savefig('border_removal_check.png', dpi=150, bbox_inches='tight')
plt.show()
```
